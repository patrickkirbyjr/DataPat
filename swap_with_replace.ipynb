# How to Replace Values in a Pandas DataFrame Using .replace()

## What is .replace()?

Used to replace values in a Pandas Series or DataFrame

Tell it what to find and give it something to replace it with

```python
# import libraries
import pandas as pd
import seaborn as sns
```

```python
# import dataset (built-in seaborn dataset)
penguins = sns.load_dataset('penguins')
penguins.head()
```

## Replacing Specific Values

```python
# print unique penguin species
print(penguins['species'].unique())
```

```python
# replace "Gentoo" species with "Northern Gentoo"
penguins['species'] = penguins['species'].replace('Gentoo', 'Northern Gentoo')
print(penguins['species'].unique())
```

```python
# see changes in dataframe
penguins.tail()
```

## Replacing Multiple Values with the Same Value

```python
# replace island names with "Deception"
penguins['island'] = penguins['island'].replace(['Biscoe', 'Dream', 'Torgersen'], 'Deception')
print(penguins['island'].unique())
```

```python
# see changes in dataframe
penguins.head()
```

## Replacing Multiple Values with Different Values

```python
# create dictionary for sex abbreviations
sex_abb = {'Male': 'M', 'Female': 'F'}

# replace penguin sex values with abbreviations
penguins['sex'] = penguins['sex'].replace(sex_abb)
penguins.head()
```

### Note: If you want to replace NA values, it is best to use .fillna()
