# Creating a Simple Line Graph with Matplotlib Part 2: Adding Lines & Ticks

better_graph.png

```python
import matplotlib.pyplot as plt #import library
```

```python
fig, ax = plt.subplots(figsize=(4,3))

ax.plot([1, 2, 3, 4], [1, 2, 3, 4])

ax.set_title('A Simple Graph')

ax.set_xlabel('X Label')
ax.set_ylabel('Y Label')

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(4,3))

ax.plot([1, 2, 3, 4], [1, 2, 3, 4])

#consolidate label code
ax.set(title='A Simple Graph', xlabel='X Label', ylabel='Y Label')

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(4,3))

ax.plot([1, 2, 3, 4], [1, 2, 3, 4])

ax.set(
    title='A Simple Graph',
    xlabel='X Label',
    ylabel='Y Label'
)

ax.grid() #show gridlines on graph

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(4,3))

ax.plot([1, 2, 3, 4], [1, 2, 3, 4])

ax.set(
    title='A Simple Graph',
    xlabel='X Label',
    ylabel='Y Label'
)

ax.set_xticks([1, 2, 3, 4]) #set x ticks
ax.set_yticks([1, 2, 3, 4]) #set y ticks

ax.grid()

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(4,3))

ax.plot([1, 2, 3, 4], [1, 2, 3, 4])

ax.set(
    title='A Simple Graph',
    xlabel='X Label',
    ylabel='Y Label'
)

ax.set_xticks([1, 2, 3, 4], labels=['One', 'Two', 'Three', 'Four']) #set x tick labels
ax.set_yticks([1, 2, 3, 4], labels=['One', 'Two', 'Three', 'Four']) #set y tick labels

ax.grid()

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(4,3))

ax.plot([1, 2, 3, 4], [1, 2, 3, 4])

ax.set(
    title='A Simple Graph',
    xlabel='X Label',
    ylabel='Y Label'
)

ax.set_xticks([1, 2, 3, 4], labels=['One', 'Two', 'Three', 'Four'])
ax.set_yticks([1, 2, 3, 4], labels=['One', 'Two', 'Three', 'Four'])

ax.grid()

plt.savefig('better_graph.png', bbox_inches='tight') #save your graph as png

plt.show()
```
