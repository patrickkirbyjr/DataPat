# SQL Window Functions: Rank in Python with DuckDB

```python
#import libraries
import duckdb
```

```python
#ingest data
url = "https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_2023-01.parquet"
df_preview = duckdb.sql(f"SELECT * FROM '{url}' LIMIT 5").df()
df_preview
```

```python
#write query for trip distance running totals for January 2023
query = f"""
  SELECT
    passenger_count,
    trip_distance,
    RANK() OVER (PARTITION BY passenger_count ORDER BY trip_distance DESC) AS distance_rank
  FROM '{url}'
  WHERE passenger_count IS NOT NULL
  QUALIFY distance_rank <= 3
  ORDER BY passenger_count
"""
df = duckdb.sql(query).df()
df
```
