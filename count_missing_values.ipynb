# How to Find and Count Missing Values (Python Colab Tutorial)



## How do I quickly find NaN or missing values within a dataset?

NaN - Not a Number

Use .isna() to check for missing values using boolean indexing

```python
# import libraries
import seaborn as sns
import pandas as pd
```

```python
# import dataset (built-in seaborn dataset)
penguins = sns.load_dataset('penguins')
penguins.head(10)
```

```python
# identify missing values
missing_values = penguins.isna()
missing_values.head(10)
```

```python
# count missing values by column
missing_counts = penguins.isna().sum()
print(missing_counts)
```

```python
# count total missing values
total_missing = penguins.isna().sum().sum()
print(total_missing)
```
