# How to Fill Missing Values in Pandas (Python Colab Tutorial)

## Identify Missing Values

```python
# import libraries
import pandas as pd
import seaborn as sns
```

```python
# import dataset (built-in seaborn dataset)
penguins = sns.load_dataset('penguins')
penguins
```

```python
# count missing values for each column
missing_counts = penguins.isna().sum()
print(missing_counts)
```

## Fill Missing Categorical Values with "Unknown"

```python
# replace missing categorical values
penguins['sex'] = penguins['sex'].fillna('Unknown')
penguins
```

## Fill Missing Numerical Values with Median/Mean

```python
# calculate median and mean values
median_bill_length = penguins['bill_length_mm'].median()
median_bill_depth = penguins['bill_depth_mm'].median()
median_flipper_length = penguins['flipper_length_mm'].median()
average_body_mass = penguins['body_mass_g'].mean()


# replace missing values calculated values
penguins['bill_length_mm'] = penguins['bill_length_mm'].fillna(median_bill_length)
penguins['bill_depth_mm'] = penguins['bill_depth_mm'].fillna(median_bill_depth)
penguins['flipper_length_mm'] = penguins['flipper_length_mm'].fillna(median_flipper_length)
penguins['body_mass_g'] = penguins['body_mass_g'].fillna(average_body_mass)
penguins
```
