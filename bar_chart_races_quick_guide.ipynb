# How to Make Bar Chart Races in Python - Quick Guide

```python
# install bar_chart_race library
!pip install bar_chart_race
```

```python
# import libraries
import bar_chart_race as bcr
import pandas as pd
from IPython.display import Video
```

```python
# import dataset (built-in bcr dataset)
df = bcr.load_dataset("covid19")
df.head()
```

```python
# create the bar chart race
bcr.bar_chart_race(
    df=df,
    filename="bar_chart_race.mp4",
    n_bars=10,
    title="Top 10 COVID-19 Cases by Country"
)
```

```python
# display the video
Video("bar_chart_race.mp4", embed=True, width=600)
```
