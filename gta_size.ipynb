# From GTA V to GTA VI: 13 Years & 210 GB Later (Game Bloat Visualized in Python)

```python
# wrangle data with pandas
import pandas as pd

game_data = {
    'title': [
        'GTA V',
        'Dragon Age: Inquisition',
        'Middle-earth: Shadow of Mordor',
        'The Witcher 3',
        'Fallout 4',
        'Battlefield 1',
        'Assassin\'s Creed Origins',
        'Red Dead Redemption 2',
        'Call of Duty: Modern Warfare',
        'Cyberpunk 2077',
        'Assassin\'s Creed Valhalla',
        'Forza Horizon 5',
        'God of War Ragnarok',
        'Elden Ring',
        'Baldur\'s Gate 3',
        'Starfield',
        'Final Fantasy VII Rebirth',
        'S.T.A.L.K.E.R. 2',
        'Assassin\'s Creed Shadows',
        'Monster Hunter Wilds',
        'Grand Theft Auto VI'
    ],
    'release_year': [
        2013, 2014, 2014, 2015, 2015, 2016, 2017, 2018, 2019, 2020,
        2020, 2021, 2022, 2022, 2023, 2023, 2024, 2024, 2025, 2025,
        2026
    ],
    'size_gb': [
        65, 45, 44, 50, 30, 50, 45, 120, 175, 70,
        50, 110, 90, 60, 150, 125, 145, 160, 103, 75,
        210
    ]
}

games_df = pd.DataFrame(game_data)
games_df
```

```python
# visualize with plotly
import plotly.express as px

fig = px.scatter(
    games_df,
    x='release_year',
    y='size_gb',
    size='size_gb',
    color='size_gb',
    hover_name='title',
    hover_data={'release_year': True, 'size_gb': True},
    trendline="ols",
    title='GTA VI & The 13-Year AAA Game File Size Explosion (2013–2026)',
    labels={'release_year': 'Release Year', 'size_gb': 'Install Size (GB)'}
)

fig.update_xaxes(dtick=1)

fig.update_layout(template='plotly_dark', width=800, height=400)
fig.show()
```
