# From GTA V to GTA VI: 13 Years & 210 GB Later (Game Bloat Visualized in Python)

```python
# wrangle data with pandas
import pandas as pd

game_data = {
    'title': [
        'GTA V',
        'Battlefield 4',
        'BioShock Infinite',
        'Destiny',
        'Far Cry 4',
        'The Witcher 3',
        'Fallout 4',
        'Battlefield 1',
        'Doom (2016)',
        'Horizon Zero Dawn',
        'Star Wars Battlefront II',
        'Red Dead Redemption 2',
        'God of War (2018)',
        'Call of Duty: Modern Warfare',
        'Cyberpunk 2077',
        'The Last of Us Part II',
        'Forza Horizon 5',
        'Halo Infinite',
        'God of War Ragnarok',
        'Elden Ring',
        'Baldur\'s Gate 3',
        'Starfield',
        'Final Fantasy VII Rebirth',
        'Call of Duty: Black Ops 6',
        'Monster Hunter Wilds',
        'Death Stranding 2: On the Beach',
        'Grand Theft Auto VI'
    ],
    'release_year': [
        2013, 2013, 2013, 2014, 2014, 2015, 2015, 2016, 2016, 2017,
        2017, 2018, 2018, 2019, 2020, 2020, 2021, 2021, 2022, 2022,
        2023, 2023, 2024, 2024, 2025, 2025, 2026
    ],
    'size_gb': [
        65, 30, 20, 40, 30, 50, 30, 50, 55, 45,
        60, 120, 45, 175, 70, 78, 110, 50, 90, 60,
        150, 125, 145, 128, 75, 150, 210
    ]
}

games_df = pd.DataFrame(game_data)
games_df
```

```python
# visualize with plotly
import plotly.express as px

fig = px.scatter(
    games_df,
    x='release_year',
    y='size_gb',
    size='size_gb',
    color='size_gb',
    hover_name='title',
    hover_data={'release_year': True, 'size_gb': True},
    trendline="ols",
    title='GTA VI & The 13-Year AAA Game File Size Explosion (2013–2026)',
    labels={'release_year': 'Release Year', 'size_gb': 'Install Size (GB)'}
)

fig.update_xaxes(dtick=1)

fig.update_layout(template='plotly_dark', width=800, height=400)
fig.show()
```
