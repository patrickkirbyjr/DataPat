# Run SQL Queries in Python with DuckDB

```python
# import libraries
import duckdb
```

```python
# show data
duckdb.sql("SELECT * FROM 'ev_stations.csv'").show()
```

```python
# query data
query = """
  SELECT *
  FROM 'ev_stations.csv'
"""

duckdb.sql(query).show()
```
