# How to Drop Columns in Pandas (Python Colab Tutorial)

## Dropping Columns

Dropping entire columns at once

Dublicates, useless columns, sensitive information

Clean and efficient

Drop one or multiple columns at once

.drop()

```python
# import libraries
import pandas as pd
import seaborn as sns
```

```python
# import dataset (built-in seaborn dataset)
penguins = sns.load_dataset('penguins')
penguins.head()
```

## Drop a Single Column

```python
# drop the island column
penguins_no_island = penguins.drop(columns='island')
penguins_no_island.head()
```

```python
# drop the body mass column
penguins_no_mass = penguins.drop(columns='body_mass_g')
penguins_no_mass.head()
```

## Drop Multiple Columns



```python
# drop measurement columns
penguins_no_measurements_ = penguins.drop(columns=['bill_length_mm', 'bill_depth_mm', 'flipper_length_mm', 'body_mass_g'])
penguins_no_measurements_.head()
```

```python
# drop measurement columns with a list
measurement_cols = ['bill_length_mm', 'bill_depth_mm', 'flipper_length_mm', 'body_mass_g']

penguins_no_measurements = penguins.drop(columns=measurement_cols)
penguins_no_measurements.head()
```
