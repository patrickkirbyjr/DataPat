# Run SQL SELECT Statements in Python with DuckDB

```python
# import libraries
import duckdb
import pandas as pd
```

```python
# ingest data
parquet_url = "https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_2023-01.parquet"
df_taxis = pd.read_parquet(parquet_url)
df_taxis
```

```python
#write SQL query
taxi_metrics_query = """
  SELECT
    passenger_count AS passengers,
    trip_distance AS distance,
    fare_amount AS fare
  FROM df_taxis
"""

taxi_metrics_df = duckdb.sql(taxi_metrics_query).df()
taxi_metrics_df
```
