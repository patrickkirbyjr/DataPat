# How to Load Built-In Datasets in Seaborn (Python Colab Tutorial)

## What are Seaborn datasets?

Seaborn = popular visualization library

Comes with ready-to-use datasets

Easy way to get data to work with

```python
# install seaborn (if not using Google Colab)
!pip install seaborn
```

```python
# import seaborn library
import seaborn as sns
```

## How do I know what datasets are in Seaborn?

```python
# see a list of available datasets
print(sns.get_dataset_names())
```

## How do I load a single dataset from Seaborn?

```python
# load one particular dataset
penguins = sns.load_dataset('penguins')
```

```python
# view chosen dataset
penguins
```
