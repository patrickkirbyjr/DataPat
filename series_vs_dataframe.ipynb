# Pandas Series vs. Dataframe (Python Colab Tutorial)

## Series vs. Dataframe

Series = 1D array (single row or column)

Dataframe = 2D table (rows and columns)

```python
# import libraries
import pandas as pd
```

## Let's look at a series.

```python
# create a series
penguin_species = pd.Series(['Adelie', 'Chinstrap', 'Gentoo'], name='species')
penguin_species
```

```python
# check data type
type(penguin_species)
```

## Let's look at a dataframe.

```python
# create a table
penguin_data = {
    'Name': ['John', 'Paul', 'George', 'Ringo'],
    'Species': ['Adelie', 'Adelie', 'Gentoo', 'Chinstrap'],
    'Age': [5, 5, 7, 4]
}

# create a dataframe from table
penguins = pd.DataFrame(penguin_data)
penguins
```

```python
# check data type
type(penguins)
```
