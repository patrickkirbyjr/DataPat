```python
import matplotlib.pyplot as plt #import library
```

```python
fig, ax = plt.subplots(figsize=(5,4)) #make plot slightly larger

ax.plot([1, 2, 3, 4], [1, 2, 3, 4], label='Original Data')
ax.plot([1, 2, 3, 4], [1, 1.5, 2, 2.5], label='Second Data')
ax.plot([1, 2, 3, 4], [1, 3, 1.5, 5], label='Third Data')

ax.axhline(y=2.5)

ax.set(
    title='A Simple Graph',
    xlabel='X Label',
    ylabel='Y Label'
)

ax.set_xticks([1, 2, 3, 4], labels=['One', 'Two', 'Three', 'Four'])
ax.set_yticks([1, 2, 3, 4], labels=['One', 'Two', 'Three', 'Four'])

ax.grid()
ax.legend()

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(5,4))

#format lines
ax.plot([1, 2, 3, 4], [1, 2, 3, 4], label='Original Data', color='#1d3557', linewidth=2)
ax.plot([1, 2, 3, 4], [1, 1.5, 2, 2.5], label='Second Data', color='#457b9d', linewidth=2)
ax.plot([1, 2, 3, 4], [1, 3, 1.5, 5], label='Third Data', color='#B072AB', linewidth=2)

ax.axhline(y=2.5)

ax.set(
    title='A Simple Graph',
    xlabel='X Label',
    ylabel='Y Label'
)

ax.set_xticks([1, 2, 3, 4], labels=['One', 'Two', 'Three', 'Four'])
ax.set_yticks([1, 2, 3, 4], labels=['One', 'Two', 'Three', 'Four'])

ax.grid()
ax.legend()

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(5,4))

ax.plot([1, 2, 3, 4], [1, 2, 3, 4], label='Original Data', color='#1d3557', linewidth=2)
ax.plot([1, 2, 3, 4], [1, 1.5, 2, 2.5], label='Second Data', color='#457b9d', linewidth=2)
ax.plot([1, 2, 3, 4], [1, 3, 1.5, 5], label='Third Data', color='#B072AB', linewidth=2)

#format reference line
target_y = 2.5
ax.axhline(y=target_y, linestyle='--', linewidth=1.2, color='#C8102E', alpha=0.6)
ax.text(1.05, target_y+0.1, 'Target (2.5)', color='#C8102E')

ax.set(
    title='A Simple Graph',
    xlabel='X Label',
    ylabel='Y Label'
)

ax.set_xticks([1, 2, 3, 4], labels=['One', 'Two', 'Three', 'Four'])
ax.set_yticks([1, 2, 3, 4], labels=['One', 'Two', 'Three', 'Four'])

ax.grid()
ax.legend()

plt.show()
```
