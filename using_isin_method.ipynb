# How to Use .isin() for Fast Row Filtering (Python Colab Tutorial)

## What is .isin()?

Used to filter with multiple conditions

Cleaner than using multiple OR (|) statements

```python
# import libraries
import seaborn as sns
import pandas as pd
```

```python
# import dataset (built-in seaborn dataset)
penguins = sns.load_dataset('penguins')
penguins.head()
```

## The Messy Way





```python
# filter for penguins that are adelie or chinstrap without .isin()
filtered_species = penguins[(penguins['species'] == 'Adelie') | (penguins['species'] == 'Chinstrap')]
filtered_species
```

## The Clean Way

```python
# filter for penguins that are adelie or chinstrap with .isin()
target_species = ['Adelie', 'Chinstrap']

filtered_penguins = penguins[penguins['species'].isin(target_species)]
filtered_penguins
```

# BONUS BIT - Inverting the Filter!

Use a simple "~" to invert the filter

```python
# filter for penguins that are NOT adelie or chinstrap with .isin()
not_adelie_chinstrap = penguins[~penguins['species'].isin(target_species)]
not_adelie_chinstrap
```
