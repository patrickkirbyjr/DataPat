# Use Plotly to Make an Interactive Box & Violin Plots

```python
# import libraries
import plotly.express as px
import seaborn as sns
```

```python
# import penguin data
df = sns.load_dataset('penguins').dropna()
df.head()
```

```python
# create box plot
fig = px.box(
    df,
    x='species',
    y='flipper_length_mm',
    color='species',
    points='all',
    title='Flipper Length Distribution by Species (Box Plot)',
    labels={
        'species': 'Species',
        'flipper_length_mm': 'Flipper Length (mm)'
    },
    width=800,
    height=400,
    template='plotly_dark'
)
fig.show()
```

```python
# create violin plot
fig = px.violin(
    df,
    x='species',
    y='flipper_length_mm',
    color='species',
    box=True,
    title='Flipper Length Distribution by Species (Violin Plot)',
    labels={
        'species': 'Species',
        'flipper_length_mm': 'Flipper Length (mm)'
    },
    width=800,
    height=400,
    template='plotly_dark'
)
fig.show()
```
