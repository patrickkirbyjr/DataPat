# How to Convert Column Names and Values to Uppercase in Pandas (Python Colab Tutorial)

# StRiNgS -> STRINGS

.rename()

.upper()

```python
# import libraries
import pandas as pd
import seaborn as sns
```

```python
# import dataset (built-in seaborn dataset)
penguins = sns.load_dataset('penguins')
```

```python
# rename columns (make them wacky)
wacky_cols = ['Species', 'iSLAND', 'Bill_Length_MM', 'BILL_DEPTH_MM', 'Flipper_Length_mm', 'body_mass_G', 'SEX']
penguins.columns = wacky_cols
penguins.head()
```

## Convert a Single Column Name to Uppercase

```python
# convert species column name to uppercase
penguins = penguins.rename(columns={
    'Species': 'SPECIES'
})
penguins.head()
```

## Convert All Column Names to Uppercase

```python
# convert column names to uppercase
penguins.columns = penguins.columns.str.upper()
penguins.head()
```

## Convert an Entire Column to Uppercase

```python
# convert species column data to uppercase
penguins['SPECIES'] = penguins['SPECIES'].str.upper()
penguins.head()
```

## Convert All Data to Uppercase

```python
# convert all data to uppercase
penguins = penguins.map(lambda x: x.upper() if isinstance(x, str) else x)
penguins.head()
```
