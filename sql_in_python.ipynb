# Filter Data in Python with SQL

```python
# import libraries
import duckdb
import pandas as pd
```

```python
# ingest data
parquet_url = "https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_2023-01.parquet"
df_taxis = pd.read_parquet(parquet_url)
df_taxis
```

```python
# filter using SQL query
query = """
SELECT
  passenger_count AS passengers,
  trip_distance AS distance,
  fare_amount AS fare,
  tip_amount AS tip
FROM df_taxis
WHERE passengers > 2.0 AND distance > 10.0
"""
filtered_df = duckdb.sql(query).df()
filtered_df
```
