# How to Convert Column Names and Values to Lowercase in Pandas (Python Colab Tutorial)

# StRiNgS -> strings

.rename()

.lower()

```python
# import libraries
import pandas as pd
import seaborn as sns
```

```python
# import dataset (built-in seaborn dataset)
penguins = sns.load_dataset('penguins')
```

```python
# rename columns (make them wacky)
wacky_cols = ['Species', 'iSLAND', 'Bill_Length_MM', 'BILL_DEPTH_MM', 'Flipper_Length_mm', 'body_mass_G', 'SEX']
penguins.columns = wacky_cols
penguins.head()
```

## Convert a Single Column Name to Lowercase

```python
# convert species column name to lowercase
penguins = penguins.rename(columns={
    'Species': 'species'
})
penguins.head()
```

## Convert All Column Names to Lowercase

```python
# convert column names to lowercase
penguins.columns = penguins.columns.str.lower()
penguins.head()
```

## Convert an Entire Column to Lowercase

```python
# convert species column data to lowercase
penguins['species'] = penguins['species'].str.lower()
penguins.head()
```

## Convert All Data to Lowercase

```python
# convert all data to lowercase (cannot use .lower)
penguins = penguins.map(lambda x: x.lower() if isinstance(x, str) else x)
penguins.head()
```
