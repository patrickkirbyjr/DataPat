# Use Plotly to Make an Interactive Scatter Plot

```python
# import libraries
import plotly.express as px
import seaborn as sns
```

```python
# import penguin data
df = sns.load_dataset('penguins')
df.head()
```

```python
# create scatter plot
fig = px.scatter(
    df,
    x='bill_length_mm',
    y='flipper_length_mm',
    color='species',
    width=800,
    height=400,
    title='Penguin Dimensions: Bill Length vs. Flipper Length by Species'
)
fig.show()
```
