# Visualizing SQL Joins in Python (Left, Right, Inner, Outer)

```python
#import libraries
import pandas as pd
```

```python
#create dataframes
members = pd.DataFrame({
    'name': ['John Lennon', 'Paul McCartney', 'George Harrison', 'Ringo Starr', 'Pete Best'],
    'role': ['Guitar', 'Bass', 'Guitar', 'Drums', 'Drums (Original)']
})

instruments = pd.DataFrame({
  'name': [
      'John Lennon', 'Paul McCartney', 'George Harrison', 'Ringo Starr',
      'Jimi Hendrix', 'Eric Clapton', 'Slash', 'Jimmy Page', 'Eddie Van Halen', 'David Gilmour'
    ],
    'instrument': [
        'Rickenbacker 325', 'Höfner 500/1', 'Gretsch Duo Jet', 'Ludwig Drum Kit',
        'Fender Stratocaster', 'Gibson Les Paul', 'Gibson Les Paul Custom',
        'Gibson EDS-1275', 'Frankenstrat', 'Fender Stratocaster'
    ]
})
```

```python
#view members dataframe
members
```

```python
#view instruments dataframe
instruments
```

```python
#left join
left = pd.merge(members, instruments, on='name', how='left')
left
```

```python
#right join
right = pd.merge(members, instruments, on='name', how='right')
right
```

```python
#inner join
inner = pd.merge(members, instruments, on='name', how='inner')
inner
```

```python
#outer join
outer = pd.merge(members, instruments, on='name', how='outer')
outer
```
