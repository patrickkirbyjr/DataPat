# Creating a Simple Line Graph with Matplotlib Part 3: Adding More Data & Labels

even_better_graph.png

```python
import matplotlib.pyplot as plt #import library
```

```python
fig, ax = plt.subplots(figsize=(4,3))

ax.plot([1, 2, 3, 4], [1, 2, 3, 4])

ax.set(
    title='A Simple Graph',
    xlabel='X Label',
    ylabel='Y Label'
)

ax.set_xticks([1, 2, 3, 4], labels=['One', 'Two', 'Three', 'Four'])
ax.set_yticks([1, 2, 3, 4], labels=['One', 'Two', 'Three', 'Four'])

ax.grid()

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(4,3))

ax.plot([1, 2, 3, 4], [1, 2, 3, 4])
ax.plot([1, 2, 3, 4], [1, 1.5, 2, 2.5]) #add a second set of data

ax.set(
    title='A Simple Graph',
    xlabel='X Label',
    ylabel='Y Label'
)

ax.set_xticks([1, 2, 3, 4], labels=['One', 'Two', 'Three', 'Four'])
ax.set_yticks([1, 2, 3, 4], labels=['One', 'Two', 'Three', 'Four'])

ax.grid()

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(4,3))

ax.plot([1, 2, 3, 4], [1, 2, 3, 4])
ax.plot([1, 2, 3, 4], [1, 1.5, 2, 2.5])
ax.plot([1, 2, 3, 4], [1, 3, 1.5, 5]) #add a third set of data

ax.set(
    title='A Simple Graph',
    xlabel='X Label',
    ylabel='Y Label'
)

ax.set_xticks([1, 2, 3, 4], labels=['One', 'Two', 'Three', 'Four'])
ax.set_yticks([1, 2, 3, 4], labels=['One', 'Two', 'Three', 'Four'])

ax.grid()

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(4,3))

ax.plot([1, 2, 3, 4], [1, 2, 3, 4], label='Original Data') #add data labels
ax.plot([1, 2, 3, 4], [1, 1.5, 2, 2.5], label='Second Data') #add data labels
ax.plot([1, 2, 3, 4], [1, 3, 1.5, 5], label='Third Data') #add data labels

ax.set(
    title='A Simple Graph',
    xlabel='X Label',
    ylabel='Y Label'
)

ax.set_xticks([1, 2, 3, 4], labels=['One', 'Two', 'Three', 'Four'])
ax.set_yticks([1, 2, 3, 4], labels=['One', 'Two', 'Three', 'Four'])

ax.grid()

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(4,3))

ax.plot([1, 2, 3, 4], [1, 2, 3, 4], label='Original Data')
ax.plot([1, 2, 3, 4], [1, 1.5, 2, 2.5], label='Second Data')
ax.plot([1, 2, 3, 4], [1, 3, 1.5, 5], label='Third Data')

ax.set(
    title='A Simple Graph',
    xlabel='X Label',
    ylabel='Y Label'
)
ax.set_xticks([1, 2, 3, 4], labels=['One', 'Two', 'Three', 'Four'])
ax.set_yticks([1, 2, 3, 4], labels=['One', 'Two', 'Three', 'Four'])

ax.grid()
ax.legend() #add legend

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(4,3))

ax.plot([1, 2, 3, 4], [1, 2, 3, 4], label='Original Data')
ax.plot([1, 2, 3, 4], [1, 1.5, 2, 2.5], label='Second Data')
ax.plot([1, 2, 3, 4], [1, 3, 1.5, 5], label='Third Data')

ax.set(
    title='A Simple Graph',
    xlabel='X Label',
    ylabel='Y Label'
)
ax.set_xticks([1, 2, 3, 4], labels=['One', 'Two', 'Three', 'Four'])
ax.set_yticks([1, 2, 3, 4], labels=['One', 'Two', 'Three', 'Four'])

ax.grid()
ax.legend()

plt.savefig('even_better_graph.png', bbox_inches='tight') #save your graph as png

plt.show()
```
