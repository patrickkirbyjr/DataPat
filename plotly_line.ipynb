# Use Plotly to Make an Interactive Line Graph

```python
# import libraries
import plotly.express as px
import seaborn as sns
```

```python
# import dow jones data
df = sns.load_dataset('dowjones')
df.head()
```

```python
# create line graph
fig = px.line(
    df,
    x='Date',
    y='Price',
    title='Dow Jones Industrial Average (USD)',
    width=700,
    height=350,
    template='plotly_dark'
)
fig.show()
```
