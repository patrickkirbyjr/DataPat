# How to Inspect a Dataframe: Content Summaries (Python Colab Tutorial)

## What are content summaries of a dataframe?

Essentially, you're taking a look at basic statistics of the dataframe

Minimum, maximum, mean, standard deviation, and other info

```python
# import libraries
import seaborn as sns
import pandas as pd
```

```python
# import dataset (built-in seaborn dataset)
penguins = sns.load_dataset('penguins')
```

```python
# show dataset
penguins
```

## Method #1: .info()

```python
# summarize data with info
penguins.info()
```

## Method #2: .describe()

```python
# summarize dataframe with describe
penguins.describe()
```
