# How to Filter Ranges in Pandas Using .between()

❌ BAD:

penguins[(penguins['flipper_length_mm'] >= 190) & (penguins['flipper_length_mm'] <= 210)]

✅ GOOD:

penguins[penguins['flipper_length_mm'].between(190, 210)

⚙️ PARAMETERS:

1. Left boundary
2. Right boundary
3. Inclusivity - both (default), neither, left, right

```python
# import libraries
import pandas as pd
import seaborn as sns
```

```python
# import dataset (built-in seaborn dataset)
penguins = sns.load_dataset('penguins')
penguins
```

```python
# get penguins with medium flipper lengths - default exclusivity (both)
medium_flipper_penguins = penguins[penguins['flipper_length_mm'].between(190,210)]
medium_flipper_penguins
```

```python
# get penguins with medium flipper lengths - inclusivity both
medium_flipper_penguins = penguins[penguins['flipper_length_mm'].between(190,210,'both')]
medium_flipper_penguins
```

```python
# get penguins with medium flipper lengths - inclusivity left
medium_flipper_penguins = penguins[penguins['flipper_length_mm'].between(195,210, 'left')]
medium_flipper_penguins
```

```python
# get penguins with medium flipper lengths - inclusivity right
medium_flipper_penguins = penguins[penguins['flipper_length_mm'].between(195,210,'right')]
medium_flipper_penguins
```

```python
# get penguins with medium flipper lengths - inclusivity neither
medium_flipper_penguins = penguins[penguins['flipper_length_mm'].between(190,210, 'neither')]
medium_flipper_penguins
```
