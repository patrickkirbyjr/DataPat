# Creating a Simple Line Graph with Matplotlib Part 1: Making the Graph



first_graph.png

```python
import matplotlib.pyplot as plt #import library
```

```python
fig, ax = plt.subplots() #create figure canvas (fig) and plotting area (ax)
```

```python
fig, ax = plt.subplots(figsize=(4,3)) #change size
```

```python
fig, ax = plt.subplots(figsize=(4,3))

ax.plot([1, 2, 3, 4], [1, 2, 3, 4]) #plot data
```

```python
fig, ax = plt.subplots(figsize=(4,3))

ax.plot([1, 2, 3, 4], [1, 2, 3, 4])

plt.show() #show the graph (optional in Colab)
```

```python
fig, ax = plt.subplots(figsize=(4,3))

ax.plot([1, 2, 3, 4], [1, 2, 3, 4])

ax.set_title('A Simple Graph') #add title

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(4,3))

ax.plot([1, 2, 3, 4], [1, 2, 3, 4])

ax.set_title('A Simple Graph')

ax.set_xlabel('X Label') # add x label

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(4,3))

ax.plot([1, 2, 3, 4], [1, 2, 3, 4])

ax.set_title('A Simple Graph')

ax.set_xlabel('X Label')
ax.set_ylabel('Y Label') #add y label

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(4,3))

ax.plot([1, 2, 3, 4], [1, 2, 3, 4])

ax.set_title('A Simple Graph')

ax.set_xlabel('X Label')
ax.set_ylabel('Y Label')

plt.savefig('first_graph.png', bbox_inches='tight') #save your graph as png

plt.show()
```
