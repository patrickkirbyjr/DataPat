# How to Rename Columns in Pandas (Python Colab Tutorial)

## Renaming Columns

Important first step in making sense of data

Fix typos and make columns easier to work with

.rename() makes this easy!

```python
# import libraries
import pandas as pd
import seaborn as sns
```

```python
# import dataset (built-in seaborn dataset)
penguins = sns.load_dataset('penguins')
penguins.head()
```

## Renaming Specific Columns

```python
# rename specific columns using a dictionary
penguins_renamed = penguins.rename(columns={
    'species': 'penguin_species',
    'island': 'origin_island'
})
penguins_renamed.head()
```

## Renaming Every Column

```python
# create a list of new column names
new_columns = ['Species', 'Island', 'Bill_Length', 'Bill_Depth', 'Flipper_Length', 'Body_Mass', 'Sex']

# assign new names
penguins_renamed.columns = new_columns
penguins_renamed.head()
```
